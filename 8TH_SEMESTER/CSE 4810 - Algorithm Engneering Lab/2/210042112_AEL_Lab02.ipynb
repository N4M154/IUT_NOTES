```python
# 210042112
# -_- N4M154 -_-

# from helper code
class Node:
    def __init__(self, x):
        self.data = x
        self.left = None
        self.right = None

def insert(root, key):
    if root is None:
        return Node(key)
    if root.data == key:
        return root
    if root.data < key:
        root.right = insert(root.right, key)
    else:
        root.left = insert(root.left, key)
    return root


def FindNodesInRange(root, low, high, result):
    if root is None:
        return

    if root.data >= low and root.data <= high:

        FindNodesInRange(root.left, low, high, result)
        result.append(root.data)
        FindNodesInRange(root.right, low, high, result)

    elif root.data < low:

        FindNodesInRange(root.right, low, high, result)

    else:

        FindNodesInRange(root.left, low, high, result)

if __name__ == '__main__':
    ar = [11, 2, 29, 1, 7, 15, 40, 35]
    root = None
    for el in ar:
        root = insert(root, el)

    result = []
    FindNodesInRange(root, 12, 35, result)
    print(", ".join(map(str, result)))
```

```python
# 210042112
# -_- N4M154 -_-

# Helper code
class Node:
    def __init__(self, x):
        self.data = x
        self.left = None
        self.right = None

def insert(root, key):
    if root is None:
        return Node(key)
    if root.data == key:
        return root
    if root.data < key:
        root.right = insert(root.right, key)
    else:
        root.left = insert(root.left, key)
    return root


def max_sum(root):
    if root is None:
        return

    left_include, left_exclude = max_sum(root.left)
    right_include, right_exclude = max_sum(root.right)

    include = root.data + left_exclude + right_exclude

    exclude = max(left_include, left_exclude) + max(right_include, right_exclude)

    return (include, exclude)

if __name__ == '__main__':
    ar = [11, 2, 29, 1, 7, 15, 40, 35]
    root = None
    for el in ar:
        root = insert(root, el)

    include, exclude = max_sum(root)
    print(max(include, exclude))
```
