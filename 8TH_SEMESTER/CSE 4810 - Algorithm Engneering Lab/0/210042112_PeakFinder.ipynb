```python
# 210042112
# -_- N4M154 -_-

def findPeak2D(matrix):
    rows = len(matrix)
    cols = len(matrix[0])

    # Function to find the max element in a given column
    def findMaxInColumn(col):
        max_val = float('-inf')
        max_row = -1
        for row in range(rows):
            if matrix[row][col] > max_val:
                max_val = matrix[row][col]
                max_row = row
        return max_row

    # Recursive function for peak finding
    def peakHelper(left, right):
        if left > right:
            return None

        mid = (left + right) // 2
        max_row = findMaxInColumn(mid)

        # Check if the middle column element is a peak
        if (mid == 0 or matrix[max_row][mid] >= matrix[max_row][mid - 1]) and \
           (mid == cols - 1 or matrix[max_row][mid] >= matrix[max_row][mid + 1]):
            return (max_row, mid)

        # If the left neighbor is greater, move to the left half
        elif mid > 0 and matrix[max_row][mid] < matrix[max_row][mid - 1]:
            return peakHelper(left, mid - 1)

        # Otherwise, move to the right half
        else:
            return peakHelper(mid + 1, right)

    return peakHelper(0, cols - 1)

# Input matrix from the user
def get_input_matrix():
    print("Enter the number of rows:")
    rows = int(input())
    print("Enter the number of columns:")
    cols = int(input())

    matrix = []
    print("Enter the matrix row by row:")
    for i in range(rows):
        row = list(map(int, input().split()))
        matrix.append(row)

    return matrix

def display_output(matrix):
    peak = findPeak2D(matrix)
    if peak:
        row, col = peak
        print(f"Peak found at position: ({row}, {col}) with value: {matrix[row][col]}")
    else:
        print("No peak found.")


def main():
    matrix = get_input_matrix()
    display_output(matrix)


main()
```
