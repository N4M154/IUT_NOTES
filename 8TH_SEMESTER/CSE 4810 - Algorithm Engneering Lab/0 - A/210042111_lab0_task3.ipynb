```python
def find_peak_2d(matrix):
    rows = len(matrix)
    cols = len(matrix[0])
    lo, hi = 0, cols - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        best_row = max(range(rows), key=lambda r: matrix[r][mid])
        left  = matrix[best_row][mid-1] if mid > 0        else float('-inf')
        right = matrix[best_row][mid+1] if mid < cols-1   else float('-inf')
        val   = matrix[best_row][mid]
        if val >= left and val >= right:
            return (best_row, mid)
        elif left > val:
            hi = mid - 1
        else:
            lo = mid + 1
    return None

def get_input_matrix():
    rows = int(input("Enter number of rows: "))
    cols = int(input("Enter number of cols: "))
    matrix = []
    print("Enter matrix row by row:")
    for i in range(rows):
        row = list(map(int, input().split()))
        matrix.append(row)
    return matrix

def display_output(matrix):
    peak = find_peak_2d(matrix)
    if peak:
        row, col = peak
        print(f"Peak found at: ({row}, {col}) with value: {matrix[row][col]}")
    else:
        print("No peak found.")

matrix = get_input_matrix()
display_output(matrix)
```
