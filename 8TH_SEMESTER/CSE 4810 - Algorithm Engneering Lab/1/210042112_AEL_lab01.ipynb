# Task 1

```python
def next_greater_element(arr):
    n = len(arr)
    result = [-1] * n
    stack = []

    for i in range(n): # n times

        while stack and arr[i] > arr[stack[-1]]:
            idx = stack.pop()
            result[idx] = arr[i]
        stack.append(i)


    return result


# Test cases from the task pdf
print(next_greater_element([1, 3, 2, 4]))
print(next_greater_element([6, 8, 0, 1, 3]))
```

```python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if not self.empty():
            return self.items.pop()
        else:
            return None

    def empty(self):
        return len(self.items) == 0

    def size(self):
        return len(self.items)


class Deque:
    def __init__(self):
        self.items = []

    def append(self, item):
        self.items.append(item)

    def appendleft(self, item):
        self.items.insert(0, item)

    def pop(self):
        if not self.empty():
            return self.items.pop()
        return None

    def popleft(self):
        if not self.empty():
            return self.items.pop(0)
        return None

    def empty(self):
        return len(self.items) == 0

    def size(self):
        return len(self.items)

    def __str__(self):
        return str(self.items)


class StackUsingTwoQueues:
    def __init__(self):
        self.q1 = Deque()
        self.q2 = Deque()

    def push(self, x):

        self.q2.append(x)


        while not self.q1.empty():
            self.q2.append(self.q1.popleft())


        self.q1, self.q2 = self.q2, self.q1

    def pop(self):
        if self.q1.empty():
            return None
        return self.q1.popleft()

    def peek(self):
        if self.q1.empty():
            return None
        return self.q1.items[0]

    def is_empty(self):
        return self.q1.empty()

    def size(self):
        return self.q1.size()
```
